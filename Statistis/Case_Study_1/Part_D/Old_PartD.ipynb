### 🟩 <span style="color:#008000"><b>Point 1 :</b></span> Report the mean, median and standard deviation of final score for each group.

```python

#! suppose the long commutes starts from 1.5 hours 
#* Then divide students based on commute minutes and calculate the mean of final scores for each

long_commutes_filt, regular_commutes_filt = df["commute_minutes"] > 90 , df["commute_minutes"] < 90

long_stats = df[long_commutes_filt]["final_score"].agg(
    mean="mean",
    median="median",
    std="std"
)

regular_stats = df[regular_commutes_filt]["final_score"].agg(
    mean="mean",
    median="median",
    std="std"
)

stats = pd.DataFrame({
    "Long Commutes": long_stats,
    "Regular Commutes": regular_stats
})

print(stats)
```

```python
import matplotlib.pyplot as plt

stats.plot(
    kind="bar",
    figsize=(8, 6)
)

plt.title("Final Score Statistics by Commute Group")
plt.xlabel("Statistic")
plt.ylabel("Final Score")
plt.xticks(rotation=0)
plt.grid(axis="y", alpha=0.3)

plt.legend(
    title="Commute Group",
    loc="center",
    bbox_to_anchor=(0.5, 1.12),
    ncol=2
)

plt.tight_layout()
plt.show()
```

```python
branch_summary = (
    df.groupby("branch")
    .agg(
        commute_mean=("commute_minutes", "mean"),
        student_count=("commute_minutes", "count")
    )
    .reset_index()
    .sort_values("commute_mean", ascending=False)
)

plt.figure(figsize=(8, 5))

ax = sns.barplot(
    data=branch_summary,
    x="branch",
    y="commute_mean",
    hue="branch",
    palette="viridis",
    legend=False
)

# Display number of students above each bar
for i, row in branch_summary.iterrows():
    ax.text(
        list(branch_summary["branch"]).index(row["branch"]),
        row["commute_mean"] + 1,
        f'n = {row["student_count"]}',
        ha="center"
    )

plt.title("Average Commute Time by Branch")
plt.xlabel("Branch")
plt.ylabel("Average Commute Time (Minutes)")

plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

<b>⚠️ Validation Step</b><br><br>

To determine whether these explanations are supported by the data, we should also examine the <b>number of study hours</b> across commute duration and laptop-ownership groups.

</div>

```python
import matplotlib.pyplot as plt
import numpy as np

# Calculate means and counts
long_summary = df[long_commutes_filt].groupby("has_laptop")[
    ["sleep_hours", "study_hours_week"]
].agg(["mean", "count"])

regular_summary = df[regular_commutes_filt].groupby("has_laptop")[
    ["sleep_hours", "study_hours_week"]
].agg(["mean", "count"])


fig, axes = plt.subplots(1, 2, figsize=(14, 5))

width = 0.35


# ================= LONG COMMUTES =================

x = np.arange(len(long_summary))

sleep_bars = axes[0].bar(
    x - width / 2,
    long_summary["sleep_hours"]["mean"],
    width,
    label="Sleep Hours"
)

study_bars = axes[0].bar(
    x + width / 2,
    long_summary["study_hours_week"]["mean"],
    width,
    label="Study Hours / Week"
)

# Add number of students above each category
for i, category in enumerate(long_summary.index):

    count = df[long_commutes_filt][
        df[long_commutes_filt]["has_laptop"] == category
    ].shape[0]

    max_height = max(
        long_summary.loc[category, ("sleep_hours", "mean")],
        long_summary.loc[category, ("study_hours_week", "mean")]
    )

    axes[0].text(
        i,
        max_height + 0.5,
        f"n = {count}",
        ha="center",
        fontweight="bold"
    )

axes[0].set_xticks(x)
axes[0].set_xticklabels(long_summary.index.astype(str))
axes[0].set_title("Long Commutes")
axes[0].set_xlabel("Has Laptop")
axes[0].set_ylabel("Mean Hours")
axes[0].grid(axis="y", linestyle="--", alpha=0.3)


# ================= REGULAR COMMUTES =================

x = np.arange(len(regular_summary))

sleep_bars = axes[1].bar(
    x - width / 2,
    regular_summary["sleep_hours"]["mean"],
    width,
    label="Sleep Hours"
)

study_bars = axes[1].bar(
    x + width / 2,
    regular_summary["study_hours_week"]["mean"],
    width,
    label="Study Hours / Week"
)

# Add number of students above each category
for i, category in enumerate(regular_summary.index):

    count = df[regular_commutes_filt][
        df[regular_commutes_filt]["has_laptop"] == category
    ].shape[0]

    max_height = max(
        regular_summary.loc[category, ("sleep_hours", "mean")],
        regular_summary.loc[category, ("study_hours_week", "mean")]
    )

    axes[1].text(
        i,
        max_height + 0.5,
        f"n = {count}",
        ha="center",
        fontweight="bold"
    )

axes[1].set_xticks(x)
axes[1].set_xticklabels(regular_summary.index.astype(str))
axes[1].set_title("Regular Commutes")
axes[1].set_xlabel("Has Laptop")
axes[1].set_ylabel("Mean Hours")
axes[1].grid(axis="y", linestyle="--", alpha=0.3)


# ================= GENERAL DECORATION =================

fig.suptitle(
    "Sleep and Study Hours by Commute Group and Laptop Availability",
    fontsize=14,
    fontweight="bold"
)

# One legend centered at the middle/top of the figure
handles, labels = axes[0].get_legend_handles_labels()

fig.legend(
    handles,
    labels,
    loc="upper center",
    bbox_to_anchor=(0.5, 0.93),
    ncol=2
)

plt.tight_layout(rect=[0, 0, 1, 0.88])

plt.show()
```

<b>1. Long commute + No laptop:</b> <span style="color:#C00000;"><b>❌ Not supported</b></span><br>
These students study hard and still have sufficient sleeping time.<br><br>

<b>2. Long commute + Laptop:</b> <span style="color:#548235;"><b>✅ Supported</b></span><br>

<b>⚠️ Next Validation Step</b><br><br>

Add the <b>final score</b> for each category to determine whether sacrificing sleeping time for studying at home actually has a measurable effect on students' final scores.

```python
import matplotlib.pyplot as plt
import numpy as np

# Calculate means
long_summary = df[long_commutes_filt].groupby("has_laptop")[
    ["sleep_hours", "study_hours_week", "final_score"]
].mean()

regular_summary = df[regular_commutes_filt].groupby("has_laptop")[
    ["sleep_hours", "study_hours_week", "final_score"]
].mean()


fig, axes = plt.subplots(1, 2, figsize=(15, 6))

width = 0.25


# ================= LONG COMMUTES =================

x = np.arange(len(long_summary))

bars1 = axes[0].bar(
    x - width,
    long_summary["sleep_hours"],
    width,
    label="Sleep Hours"
)

bars2 = axes[0].bar(
    x,
    long_summary["study_hours_week"],
    width,
    label="Study Hours / Week"
)

bars3 = axes[0].bar(
    x + width,
    long_summary["final_score"],
    width,
    label="Final Score"
)

axes[0].set_xticks(x)
axes[0].set_xticklabels(long_summary.index.astype(str))

axes[0].set_title("Long Commutes")
axes[0].set_xlabel("Has Laptop")
axes[0].set_ylabel("Mean Value")

axes[0].grid(axis="y", linestyle="--", alpha=0.3)


# Add number of students
for i, category in enumerate(long_summary.index):

    count = (
        df[long_commutes_filt]["has_laptop"] == category
    ).sum()

    axes[0].text(
        i,
        max(
            long_summary.loc[category, "sleep_hours"],
            long_summary.loc[category, "study_hours_week"],
            long_summary.loc[category, "final_score"]
        ) + 1,
        f"n = {count}",
        ha="center",
        fontweight="bold"
    )


# ================= REGULAR COMMUTES =================

x = np.arange(len(regular_summary))

bars1 = axes[1].bar(
    x - width,
    regular_summary["sleep_hours"],
    width,
    label="Sleep Hours"
)

bars2 = axes[1].bar(
    x,
    regular_summary["study_hours_week"],
    width,
    label="Study Hours / Week"
)

bars3 = axes[1].bar(
    x + width,
    regular_summary["final_score"],
    width,
    label="Final Score"
)

axes[1].set_xticks(x)
axes[1].set_xticklabels(regular_summary.index.astype(str))

axes[1].set_title("Regular Commutes")
axes[1].set_xlabel("Has Laptop")
axes[1].set_ylabel("Mean Value")

axes[1].grid(axis="y", linestyle="--", alpha=0.3)


# Add number of students
for i, category in enumerate(regular_summary.index):

    count = (
        df[regular_commutes_filt]["has_laptop"] == category
    ).sum()

    axes[1].text(
        i,
        max(
            regular_summary.loc[category, "sleep_hours"],
            regular_summary.loc[category, "study_hours_week"],
            regular_summary.loc[category, "final_score"]
        ) + 1,
        f"n = {count}",
        ha="center",
        fontweight="bold"
    )


# ================= GENERAL DECORATION =================

fig.suptitle(
    "Sleep, Study Hours and Final Score by Commute Group and Laptop Availability",
    fontsize=14,
    fontweight="bold"
)

# One centered legend
handles, labels = axes[0].get_legend_handles_labels()

fig.legend(
    handles,
    labels,
    loc="upper center",
    bbox_to_anchor=(0.5, 0.93),
    ncol=3
)

plt.tight_layout(rect=[0, 0, 1, 0.88])

plt.show()
```
