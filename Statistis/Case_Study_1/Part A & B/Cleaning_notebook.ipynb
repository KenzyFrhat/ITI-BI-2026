```python
import re

def clean_branch(value):
    text = str(value).lower().strip()

    # Remove emoji and punctuation, keeping Arabic/English letters and spaces
    text = re.sub(r"\s+", " ", text).strip()

    if re.search(r"cairo", text):
        return "cairo"
    if re.search(r"alexandria", text):
        return "alexandria"
    if re.search(r"mansoura|mansora", text):
        return "mansoura"

    return text

df["branch"] = df["branch"].apply(clean_branch)
df["branch"].value_counts()
```

```python
import re

def clean_labtop(value):
    text = str(value).lower().strip()

    # Remove emoji and punctuation, keeping Arabic/English letters and spaces
    text = re.sub(r"\s+", " ", text).strip()

    if re.search(r"1|y", text):
        return "yes"
    if re.search(r"n|no", text):
        return "no"

    return text

df["has_laptop"] = df["has_laptop"].apply(clean_labtop)
df["has_laptop"].value_counts()
```
